```python
import socket
import subprocess
import time
from pathlib import Path

DATA = Path(r"C:\Users\User\Desktop\Sanuth\data")
VENV_PYTHON = DATA / ".venv" / "Scripts" / "python.exe"   # the Python that has Streamlit installed
PORT = 8502                                                  # the website's address: localhost:8502


def running(port):
    """True if something is already answering on this port."""
    with socket.socket() as s:
        return s.connect_ex(("localhost", port)) == 0


if running(PORT):
    print(f"The website is already running at http://localhost:{PORT}")
else:
    log = open(DATA / "website_log.txt", "w")
    website = subprocess.Popen(
        [str(VENV_PYTHON), "-m", "streamlit", "run", str(DATA / "website.py"),
         "--server.headless", "true", "--server.port", str(PORT)],
        cwd=DATA, stdout=log, stderr=subprocess.STDOUT,
    )
    for _ in range(40):          # wait up to 40 seconds for it to start
        if running(PORT):
            break
        time.sleep(1)
    if running(PORT):
        print(f"Website running at http://localhost:{PORT}")
    else:
        print("It hasn't started yet. Wait a few seconds and run this cell again, or open website_log.txt.")
```

```python
from IPython.display import IFrame, display, Markdown

display(Markdown(f"[Open the website in a new tab](http://localhost:{PORT})"))
IFrame(f"http://localhost:{PORT}", width="100%", height=900)
```

```python
try:
    website.terminate()
    print("Website stopped.")
except NameError:
    print("This notebook didn't start it (or the kernel was restarted). "
          "Closing Jupyter stops it, or restart your computer.")
```
